```python
import os
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing import image

# LOAD MODEL
model = tf.keras.models.load_model(
    "/Users/jmdhanyakumar/Desktop/MPROJECT/models/tomato_disease_model.h5"
)

# CLASS NAMES
class_names = [
    "Bacterial_Spot",
    "Cercospora_leaf_mold",
    "Early_Blight",
    "Healthy",
    "Insect_Damage",
    "Late_Blight",
    "Leaf_Miner",
    "Leaf_Mold",
    "Spider_Mites",
    "Tomato_Leaf_Curl_Virus"
]

# SELECT A FOLDER
folder = "/Users/jmdhanyakumar/Desktop/MPROJECT/dataset/Spider_Mites"

# REMOVE .DS_Store FILES
files = [
    f for f in os.listdir(folder)
    if f.lower().endswith(('.jpg', '.jpeg', '.png'))
]

# PICK FIRST IMAGE
filename = files[0]

# FULL IMAGE PATH
img_path = os.path.join(folder, filename)

print("Testing Image:", img_path)

# LOAD IMAGE
img = image.load_img(
    img_path,
    target_size=(224,224)
)

# PREPROCESS
img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0)

# PREDICT
prediction = model.predict(img_array)

predicted_class = class_names[np.argmax(prediction)]

confidence = np.max(prediction) * 100

# OUTPUT
print("\nPrediction:", predicted_class)
print("Confidence: {:.2f}%".format(confidence))
```
